Q6. Handling Missing Data (Critical ML Step)

Context: Models cannot handle null values.

Dataset: Titanic

Tasks:

1. Fill missing age using median
2. Fill missing embarked using mode
3. Drop deck column
4. Explain:
Why different strategies are used for different columns

```python
import seaborn as sns
import pandas as pd
df=sns.load_dataset("titanic")

#1Fil.l missing age using median

df['age'].fillna (df['age'].median(),inplace=True)

#2.Fill missing embarked using mode

df['embarked'].fillna(df['embarked'].mode()[0],inplace=True)

#3.Drop deck column
df.drop(coloumn=['deck'],inplace=True)

#4.Why different strategies are used for different columns

#age is numerical so median is more stable,embarked is categorical here values are labels,and drop coulmn we are using when there are somany missing values.If we filling the missing values it may introduce wrong information it is better to remove than meslead label
```

Q7. Feature Engineering for Model Improvement

Context: Raw data is rarely useful—features must be created.

Dataset: Titanic

Tasks:

1. Create:
family_size = sibsp + parch + 1
2. Create:
is_alone (binary feature)
3. Compute survival rate by is_alone
4. Explain:
Why engineered features improve ML models

```python
import seaborn as sns
import pandas as pd

df=sns.load_dataset("titanic")
#1.Create: family_size = sibsp + parch + 1
df.family_size = df["sibsp"] + df["parch"] + 1
#df.head()

#2.Create: is_alone (binary feature)

df['is_alone'] = (df['family_size'] == 1).astype(int)


#3.Compute survival rate by is_alone

survival_by_alone = df.groupby('is_alone')['survived'].mean()
print(survival_by_alone)
```

Q13. Custom Feature Transformation using apply()

Context: Real ML requires custom logic transformation.

Dataset: Titanic

Tasks:

1. Create age_group:
Child / Adult / Senior
2. Use .apply()
3. Compute survival rate per group
4. Interpret:
Which segment has highest survival likelihood

```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")
#Create age_group: Child / Adult / Senior

def agecategory(age):
    
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df['age_group']=df['age'].apply(agecategory)
print(df['age_group'])

survival_by_group = df.groupby('age_group')['survived'].mean()
print(survival_by_group)

#Interpret: Which segment has highest survival likelihood
children
```

Q14. Error Analysis Simulation (Model Evaluation Thinking)

Context: Understanding prediction errors is key in ML.

Tasks:

1. Generate 1000 random error values using NumPy
2. Compute:
Mean
Standard deviation
3. Identify outliers:
Using mean ± 2*std
4. Explain:
What high variance means in model performance

```python
import numpy as np
#Generate 1000 random error values using NumPy

errorvalue=np.random.randn(10000)
mean=np.mean(errorvalue)
std=np.std(errorvalue)
print(mean)
print(std)

#Identify outliers: Using mean ± 2*std

upper=mean-2*std
lower=mean+2*std
outliers=errorvalue[(errorvalue<lower)|(errorvalue>upper)]
print(outliers)

Explain: What high variance means in model performance

High variance means the model’s errors vary widely, indicating unstable predictions and poor generalization (overfitting)
```
